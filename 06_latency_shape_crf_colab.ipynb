# 06 · Speed of SEC-BERT-SHAPE + CRF (Colab, T4 GPU, ~10 min)

**Runtime → Change runtime type → T4 GPU**, then **Runtime → Run all**.

SEC-BERT-SHAPE + CRF adds Viterbi decoding on top of SEC-BERT-SHAPE, so it cannot reuse the SEC-BERT-SHAPE
speed. This notebook times **both** models in the same session (GPU and CPU), so the CRF's extra cost is
measured on identical hardware. Results go to `results/latency_shape_crf.csv` and
`results/LATENCY_SHAPE_CRF.md`; the main `results/latency.csv` is not changed.
